```python
flores={
    "tipo":["margarita","azucena","rosa"],
    "cant":[10,20,30],
    "precios":[1200,3000,1200]
}
```

```python
print("Bienvenido a mi floristería")
print("Se ofrecen los siguientes productos: \n")

# Encabezados de la tabla con alineación
print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15}")
print("-" * 46)

# Recorremos las tres listas simultáneamente usando zip
for tipo, cantidad, precio in zip(flores['tipo'], flores['cant'], flores['precios']):
    print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14}")
```

```python
# Inicializamos el carrito de compras fuera del ciclo para que no se borre al repetir el menú
compra_user = {
    'tipo': [],
    'cantidad': [],
    'precio': []
}

while True:
    print("\nQué desea hacer?")
    print(f"{1}. Comprar flores")
    print(f"{2}. Pagar")
    print(f"{3}. Generar factura")
    print(f"{4}. Salir")

    opc = input("Seleccione su opción: \n")
    match opc:
        case "1":
            #Falta colocar stock
            print("Flores disponibles> \n")
            for i, flor in enumerate(flores['tipo']):
                print(f"{i+1}. {flor.capitalize()} - cantidad: {flores['cant'][i]} - precio: ${flores['precios'][i]}")
            #Que pasa si el usuario entra y ya no hay flores disponibles
            if len(flores['tipo']) == 0:
                print("No hay flores en stock, sorry :(")
            #Quitar del carrito si compre lo que no queria
            if len(compra_user['tipo']) > 0:
                print("flores en el carrito: \n")
                for i, flor in enumerate(compra_user['tipo']):
                    print(f'{i+1}. {flor.capitalize()} - cantidad: {compra_user['cantidad'][i]} - precio: ${compra_user["precio"][i]}')
            #si solo compre 5 pero quiero retirar 6 , recordar cantidad que tiene y recordar que no puede retirar mas de lo que compro
            if len(compra_user['tipo']) > 0:
                print("Desea quitar o retirar alguna flor del carrito? (si/no)")
                quitar_opc = input().lower().strip()
                if quitar_opc == "si":
                    print("Flores en el carrito: \n")
                    for i, flor in enumerate(compra_user['tipo']):
                        print(f'{i+1}. {flor.capitalize()} - cantidad: {compra_user['cantidad'][i]} - precio: ${compra_user["precio"][i]}')
                    quitar_idx = int(input("Coloque el numero de la flor que deseas quitar del carrito"))
                    if quitar_idx > 0 and quitar_idx <= len(compra_user['tipo']):
                        # Aquie vamos a devolver la clase al inventario 
                        flor_a_devolver = compra_user['tipo'].pop(quitar_idx - 1)
                        cantidad_a_devolver = compra_user['cantidad'].pop(quitar_idx - 1)
                        precio_a_devolver = compra_user['precio'].pop(quitar_idx - 1)

                        # Devolvemos la flor al inventario
                        idx_inventario = flores['tipo'].index(flor_a_devolver)
                        flores['cant'][idx_inventario] += cantidad_a_devolver

            print("Ha seleccionado: Comprar flores\n")
            flor_opc = input("Seleccione la flor que desea comprar: ").lower().strip()
            if flor_opc in flores['tipo']:
                idx = flores['tipo'].index(flor_opc)
                flor_cant = int(input("Ingrese la cantidad que desea comprar: "))

                if flor_cant <= flores['cant'][idx]:
                    # Agregamos los valores a las listas del carrito
                    compra_user['tipo'].append(flor_opc)
                    compra_user['cantidad'].append(flor_cant)
                    compra_user['precio'].append(flores['precios'][idx])

                    # Restamos del inventario
                    flores['cant'][idx] -= flor_cant
                    print(f"¡Se han añadido {flor_cant} {flor_opc}(s) al carrito!")
                    # Removemos del inventario las flores si llegan a 0 usando .pop()
                    if flores['cant'][idx] == 0:
                        flores["tipo"].pop(idx)
                        flores["cant"].pop(idx)
                        flores["precios"].pop(idx)

                else:
                    print("Cantidad no disponible")
            else:
                print(f"La flor {flor_opc} no está disponible")

        case "2":
            
            if len(compra_user["tipo"]) > 0:
                print("Seleccione el medio de pago: \n")
                print(f"{1}. Efectivo")
                print(f"{2}. Tarjeta de crédito")
                print(f"{3}. Tarjeta de débito")
                print(f"{4}. Otro medio de pago")
                pago_opc = input("Seleccione su opción: \n")
                # si se compra por efectivo o credito hay que tener

                #si es efectivo hay que dar vuelto y registrar
                if pago_opc == "1":
                    efectivo_recibido = float(input("Ingrese la cantidad o monto del dinero registrado porfi: "))
                    # Calculamos el total multiplicando cada cantidad por su precio unitario
                    total_pagar = 0
                    for i in range(len(compra_user["tipo"])):
                        subtotal = compra_user["cantidad"][i] * compra_user["precio"][i]
                        total_pagar += subtotal
                        if efectivo_recibido >= total_pagar:
                            vuelto = efectivo_recibido - total_pagar
                            print(f"Su vuelto es: ${vuelto}")
                        else:
                            print("No tiene suficiente dinero para realizar la compra.")    
                #si es con tarjeta hay que registrar el numero de tarjeta y titular y aparte revisar si hay dinero en la tarjeta, si no tiene dionero suficiente no se puede hacer la compra
                if pago_opc == "2" or pago_opc == "3":
                    tarjeta_numero = input("Ingrese el número de su tarjeta: ")
                    tarjeta_titular = input("Ingrese el nombre del titular de la tarjeta: ")
                    # 
                    # Por simplicidad, asumiremos que siempre tiene fondos suficientes en este ejemplo
                # Calculamos el total multiplicando cada cantidad por su precio unitario
                total_pagar = 0
                for i in range(len(compra_user["tipo"])):
                    subtotal = compra_user["cantidad"][i] * compra_user["precio"][i]
                    total_pagar += subtotal

                print(f"\nUsted ha seleccionado la opción de pago {pago_opc}.")
                print(f"El total a pagar por su compra es: ${total_pagar}")

            else:
                print("El carrito de compras está vacío. ")

        case "3":
            print("Ha seleccionado: Generar factura\n")
            # Encabezados de la tabla con alineación
            print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15} | {'Subtotal':<15}")
            print("-" * 46)
            for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
              print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14} | ${precio*cantidad:<14} ")
            break
        case "4":
            print("Ha seleccionado: Salir\n")
            break
        case _:
            print('Número incorrecto, por favor seleccione uno correcto')

print("Gracias por emplear nuestros servicios. Feliz día")
```
